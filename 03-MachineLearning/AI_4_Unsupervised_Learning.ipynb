- Clustering is a machine learning technique used to group data based on similarity, with applications in customer segmentation and anomaly detection.
- K-means clustering partitions data into clusters based on the distance between data points and centroids but struggles with imbalanced or non-convex clusters.
- Heuristic methods such as silhouette analysis, the elbow method, and the Davies-Bouldin Index help assess k-means performance.

- DBSCAN is a density-based algorithm that creates clusters based on density and works well with natural, irregular patterns.

- HDBSCAN is a variant of DBSCAN that does not require parameters and uses cluster stability to find clusters.

- Hierarchical clustering can be divisive (top-down) or agglomerative (bottom-up) and produces a dendrogram to visualize the cluster hierarchy.

- Dimension reduction simplifies data structure, improves clustering outcomes, and is useful in tasks such as face recognition (using eigenfaces).

- Clustering and dimension reduction work together to improve model performance by reducing noise and simplifying feature selection.

- PCA, a linear dimensionality reduction method, minimizes information loss while reducing dimensionality and noise in data.

- t-SNE and UMAP are other dimensionality reduction techniques that map high-dimensional data into lower-dimensional spaces for visualization and analysis.

## **1. Unsupervised Learning Models (Dimensionality Reduction & Clustering)**

| Model Name | Brief Description | Code Syntax |
| :--- | :--- | :--- |
| **UMAP** | **Process:** Uniform Manifold Approximation and Projection used for dimensionality reduction.<br><br>**Pros:** High performance, preserves global structure.<br>**Cons:** Sensitive to parameters.<br>**Applications:** Data visualization, feature extraction.<br><br>**Key hyperparameters:**<br>- `n_neighbors`: Controls local neighborhood size (default = 15)<br>- `min_dist`: Min distance between points in embedded space (default = 0.1)<br>- `n_components`: Dimensionality of embedding (default = 2) | `from umap.umap_ import UMAP`<br><br>`umap = UMAP(n_neighbors=15, min_dist=0.1, n_components=2)` |
| **t-SNE** | **Process:** t-Distributed Stochastic Neighbor Embedding is a nonlinear dimensionality reduction technique.<br><br>**Pros:** Good for visualizing high-dimensional data.<br>**Cons:** Computationally expensive, prone to overfitting.<br>**Applications:** Data visualization, anomaly detection.<br><br>**Key hyperparameters:**<br>- `n_components`: Number of dimensions for output (default = 2)<br>- `perplexity`: Balances local and global attention (default = 30)<br>- `learning_rate`: Step size during optimization (default = 200) | `from sklearn.manifold import TSNE`<br><br>`tsne = TSNE(n_components=2, perplexity=30, learning_rate=200)` |
| **PCA** | **Process:** Principal Component Analysis is used for linear dimensionality reduction.<br><br>**Pros:** Easy to interpret, reduces noise.<br>**Cons:** Linear, may lose information in nonlinear data.<br>**Applications:** Feature extraction, compression.<br><br>**Key hyperparameters:**<br>- `n_components`: Number of principal components to retain (default = 2)<br>- `whiten`: Whether to scale the components (default = False)<br>- `svd_solver`: Algorithm to compute components (default = 'auto') | `from sklearn.decomposition import PCA`<br><br>`pca = PCA(n_components=2)` |
| **DBSCAN** | **Process:** Density-Based Spatial Clustering of Applications with Noise.<br><br>**Pros:** Identifies outliers, does not require the number of clusters.<br>**Cons:** Difficult with varying density clusters.<br>**Applications:** Anomaly detection, spatial data clustering.<br><br>**Key hyperparameters:**<br>- `eps`: Max distance between two points to be neighbors (default = 0.5)<br>- `min_samples`: Min samples in a neighborhood to form a cluster (default = 5) | `from sklearn.cluster import DBSCAN`<br><br>`dbscan = DBSCAN(eps=0.5, min_samples=5)` |
| **HDBSCAN** | **Process:** Hierarchical DBSCAN improves on DBSCAN by handling varying density clusters.<br><br>**Pros:** Better handling of varying densities.<br>**Cons:** Can be slower than DBSCAN.<br>**Applications:** Large datasets, complex clustering problems.<br><br>**Key hyperparameters:**<br>- `min_cluster_size`: Minimum size of clusters (default = 5)<br>- `min_samples`: Min samples to form a cluster (default = 10) | `import hdbscan`<br><br>`clusterer = hdbscan.HDBSCAN(min_cluster_size=5)` |
| **K-Means clustering** | **Process:** A centroid-based clustering algorithm that groups data into k clusters.<br><br>**Pros:** Efficient, simple to implement.<br>**Cons:** Sensitive to initial cluster centroids.<br>**Applications:** Customer segmentation, pattern recognition.<br><br>**Key hyperparameters:**<br>- `n_clusters`: Number of clusters (default = 8)<br>- `init`: Centroid initialization method ('k-means++' or 'random')<br>- `n_init`: Number of times algorithm runs with different seeds (default = 10) | `from sklearn.cluster import KMeans`<br><br>`kmeans = KMeans(n_clusters=3)` |
