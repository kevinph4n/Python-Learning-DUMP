### APIs
- Application Programming Interface (API): a software intermediary that allows two different applications to talk to each other
- Ex1: Logging in with Facebook/Google: When a website allows you to "Log in with Google," it uses Google’s API to verify your identity without accessing your Google password.
- Ex2: Checking the Weather: When you look at a weather app on your phone, the app uses an API to fetch real-time weather data from a remote weather service (like NOAA or Weather.com).

````python
# for example
import pandas as pd
dict = {'a':[1, 2, 3], 'b':[4, 5, 6]}
df = pd.DataFrame(dict)
# the data in dict passed to pandas API
# the DataFrame is used to communicate with the API
df.head() # when you call the method head -> df communicates with the API
# the API returns the value right below (output)
````

#### REST APIs (REprensentational State Transfer APIs)
- Another popular type of API: Allows you to communicate through the internet, take advantage of resources, data,...
- are used to interact with web services
- **Client inputs -> API to Web service -> Web service outputs -> Client**

#### HTTP (HyperText Transfer Protocol) Protocol
- REST APIs function is recalled by sending a request
- The request is communicated via HTTP message (usually contains a JSON file)
- **You/client -> Request -> Web Server**
- **Web Server -> Response -> You/client**

#### URL (Uniform Resource Locator): The most popular way to find resources on the web
- **Scheme**: the protocol (http://)
- **Internet address or Base URL**: used to find the location, for example: www.facebook.com
- **Route**: location on the web server, for example: /profile_username/
-> Altogether: http://www.facebook.com/profile_username/


#### Request and Response


**Request:**
- **Request start line:** /Get/index.html HTTP/1.0
- **Request header:** User-Agent: python-requests/2.21.0 Accept-Encoding:gzip, deflate:
- **Request Body**: some requests have a body 

**Response:**
- **Response Start line**: HTTP/1.0 200 OK
- **Response Header**: Server: Apache-Cache: UNCACHEABLE
- **Response Body**:
```
<!DOCTYPE html>
<html>
<body>
<h1>My 1st heading </h1>
<p> My 1st paragraph </p>
</body>
</html>
```


#### HTTP Methods
- **GET**: Retrieves data from the server
- **POST**: Submits data to server
- **PUT**: Updates data already on server
- **DELETE**: Deletes data from server
