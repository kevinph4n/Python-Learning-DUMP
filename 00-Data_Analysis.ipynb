# **Accessing Databases with Python**
## **VIII. Concepts of the Python DB API**
1. **Connection Objects**: Database connections, Manage transactions
2. **Cursor Objects**: Database queries

### Some of the connection mathods
- cursor()
- commit()
- rollback()
- close()


```python
# Example for learning
import sqlite3

# 1. Create a connection object (:memory: creates a temporary database in RAM)
connection = sqlite3.connect(":memory:")

# 2. Create a cursor object
cursor = connection.cursor()

# 3. Create a table and insert sample data (so the SELECT query has data to read)
cursor.execute("CREATE TABLE mytable (id INTEGER, name TEXT)")
cursor.execute("INSERT INTO mytable VALUES (1, 'Alice'), (2, 'Bob')")
connection.commit()  # Save the changes

# 4. Run queries (This matches your example)
cursor.execute("SELECT * FROM mytable")
results = cursor.fetchall()

# Print the results to see it work
print(results)

# 5. Free resources
cursor.close()
connection.close()
```
