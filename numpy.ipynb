### One Dimensional Numpy
-  (short for Numerical Python) is the foundational library for scientific computing in Python. It provides a highly efficient multidimensional array object (ndarray) and a vast collection of functions to perform fast mathematical, statistical, and algebraic operations on these arrays.



---
#### The basics and Array creation 

```python
# type()
import numpy as np
x = np.array ([1, 3, 5, 7, 9])
type(x)
```

```python
# .dtype 
import numpy as np
x = np.array ([1, 3, 5, 7, 9])
x.dtype
```

```python
# .size
import numpy as np
x = np.array ([1, 3, 5, 7, 9])
x.size
```

```python
# .ndim (the number of array dimensions)
import numpy as np
x = np.array ([1, 3, 5, 7, 9])
x.ndim
```

#### Indexing and Slicing

```python
# change the index numbers in the array
import numpy as np
x = np.array([11, 2, 4, 5])
x[0] = 22
print(x) # 11 -> 22
```

```python
# slicing
import numpy as np
x = np.array ([1, 3, 5, 7, 9])
y = x[1:4]
print(y)
```

```python
# slicing + changing the index numbers 
import numpy as np
x = np.array ([1, 3, 5, 7, 9])
x[1:4] = 100, 200, 300
print(x)
```
