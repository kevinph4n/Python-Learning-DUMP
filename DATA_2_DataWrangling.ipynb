# Data Pre-processing Techniques

## 1. **Core Concept**
* **Data Pre-processing**, often called data cleaning or data wrangling, is a necessary step in data analysis.
* It is the process of converting or mapping data from one raw form into another format to make it ready for further analysis.

## 2. **Key Topics in Data Pre-processing**
1. **Handling Missing Values:**
   * Identifying and handling conditions where a data entry is left empty.
2. **Data Formats:**
   * Data from different sources may be in various formats, units, or conventions.
   * Using Python Pandas to standardize values into the same format, unit, or convention.
3. **Data Normalization:**
   * Different columns of numerical data may have very different ranges, making direct comparison difficult.
   * Normalization brings all data into a similar range (focusing on centering and scaling) for more useful comparison.
4. **Data Binning:**
   * Creating bigger categories from a set of numerical values.
   * Particularly useful for comparison between groups of data.
5. **Categorical Variables:**
   * Converting categorical values into numeric variables to make statistical modeling easier.

## 3. **Basic Operations with Python Pandas**
* In Python, operations are usually performed along columns.
* Each column is a **Pandas Series**.
* Each row represents a sample (e.g., a different used car in the database).
* **Accessing a column:** Specify the column name (e.g., `symboling`, `body-style`).
* **Manipulating columns:** You can add a value to each entry of a column (e.g., adding 1 to each `symboling` entry).

```python
import pandas as pd

# 1. Create a sample DataFrame similar to the used car database mentioned
data = {
    'symboling': [0, 1, 2, -1, 3],
    'body_style': ['sedan', 'hatchback', 'wagon', 'sedan', 'convertible'],
    'make': ['alfa-romero', 'audi', 'bmw', 'chevrolet', 'dodge']
}
df = pd.DataFrame(data)

print("--- Original DataFrame ---")
print(df)
print("\n")

# 2. Access the 'symboling' column (a Pandas Series)
# You can access it using df['symboling']

# 3. Add 1 to each entry of the 'symboling' column
df['symboling'] = df['symboling'] + 1

print("--- DataFrame after adding 1 to 'symboling' ---")
print(df)
```

# **Handling Missing Values**

## 1. **The Problem of Missing Values**
* **Definition:** Occurs when no data value is stored for a feature in a particular observation.
* **Representation:** Often appears as `?`, `N/A`, `0`, blank cells, or `NaN` (Not a Number).

## 2. **Strategies for Dealing with Missing Data**
1. **Check the Source:** Ask the data collector if they can find the actual value.
2. **Drop Data:** Remove the data where the missing value is found.
   * **Drop Variable (Column):** If the whole column is mostly empty.
   * **Drop Entry (Row):** If only a few observations are missing. This is usually best to minimize impact.
3. **Replace Data (Imputation):** Prevents data waste, though it is less accurate as it relies on approximation.
   * *Numerical Variables:* Replace with the **average (mean)** of the entire column.
   * *Categorical Variables:* Replace with the **mode (most common value)** (e.g., fuel type).
4. **Educated Guessing:** Estimate the data if the gatherer has additional domain knowledge (e.g., older cars have higher losses).
5. **Do Nothing:** Leave it as missing if the observation is still useful to keep.

## 3. **Handling Missing Values in Python (Pandas)**
* **Dropping Missing Values (`dropna`):**
  * `df.dropna(axis=0)`: Drops rows containing missing values.
  * `df.dropna(axis=1)`: Drops columns containing missing values.
  * `inplace=True`: A parameter that writes the result directly back into the data frame without needing assignment.
* **Replacing Missing Values (`replace`):**
  * `df.replace(missing_value, new_value)` fills in the missing spots.
  * **Common Workflow:** First, calculate the column's mean (`df['column'].mean()`). Then, use `replace` to swap `NaN` for that mean.

```python
import pandas as pd
import numpy as np

# Tạo DataFrame mẫu giả lập theo bài giảng
df = pd.DataFrame({
    'make': ['alfa-romero', 'audi', 'bmw', 'chevrolet'],
    'normalized-losses': [np.nan, 164.0, np.nan, 104.0],
    'price': [13495.0, np.nan, 16430.0, 15000.0]
})

print("==== Original ====")
print(df)
print("\n")

# Ví dụ 1: Loại bỏ các hàng (rows) bị thiếu giá trị ở cột 'price' bằng dropna
# Chỉ định axis=0 để xóa hàng. Sử dụng subset để chỉ xóa nếu 'price' bị thiếu.
print("==== VÍ DỤ 1: XÓA DỮ LIỆU THIẾU (dropna) ====")
df_dropped = df.copy()
df_dropped.dropna(subset=['price'], axis=0, inplace=True)
print(df_dropped)
print("\n")

# Ví dụ 2: Thay thế giá trị thiếu ở cột 'normalized-losses' bằng giá trị trung bình (mean)
print("==== VÍ DỤ 2: THAY THẾ DỮ LIỆU THIẾU (replace) ====")
# Bước 1: Tính giá trị trung bình của cột
mean_loss = df['normalized-losses'].mean()
print(f"Giá trị trung bình của 'normalized-losses': {mean_loss}")

# Bước 2: Dùng replace để ghi đè NaN bằng giá trị trung bình vừa tính
df_replaced = df.copy()
df_replaced['normalized-losses'].replace(np.nan, mean_loss, inplace=True)
print(df_replaced)
```

# **Data Formatting**

## 1. **The Concept of Data Formatting**
* **Definition:** Bringing data into a common standard of expression that allows users to make meaningful comparisons.
* **Purpose:** Ensures data is consistent and easily understandable as part of dataset cleaning.
* **Example:** Standardizing varying expressions of the same entity (e.g., "N.Y.", "Ny", "NY", and "New York") into a single, uniform format.

## 2. **Standardizing Units and Conventions**
* **The Issue:** Data may be collected in units that are not applicable to your region or analysis (e.g., `miles per gallon` instead of the metric `liters per 100 kilometers`).
* **Unit Conversion:** You can easily transform units by applying a mathematical operation to an entire Pandas column (e.g., dividing 235 by each value in a `city-mpg` column).
* **Renaming Columns:** After converting values, use the `rename` method to update the column name to reflect the new unit.

## 3. **Handling Data Types**
* **The Issue:** Data types can be incorrectly assigned during import (e.g., a numerical `price` feature imported as an `object`). Incorrect types can cause models to behave strangely or treat valid data as missing.
* **Common Pandas Data Types:**
  * `object`: Letters or words (strings).
  * `int64`: Integers.
  * `float`: Real numbers.
* **Identifying Data Types (`dtypes`):**
  * Use `df.dtypes` to check the data type of every variable in the dataframe.
* **Converting Data Types (`astype`):**
  * Use `df.astype()` to convert a column from one format to the correct one (e.g., `df['price'].astype("int")` to convert an object column into integers).

```python
import pandas as pd

# Tạo DataFrame mẫu giả lập theo bài giảng (cố tình để price dạng chuỗi/object)
df = pd.DataFrame({
    'city-mpg': [21, 27, 19, 24],
    'price': ['13495', '16500', '18920', '12000'] 
})

print("==== DỮ LIỆU BAN ĐẦU ====")
print(df)
print("\n--- Kiểu dữ liệu hiện tại ---")
print(df.dtypes)
print("\n")

# Ví dụ 1: Chuyển đổi đơn vị tính (L/100km = 235 / mpg) và đổi tên cột
print("==== VÍ DỤ 1: CHUYỂN ĐỔI ĐƠN VỊ VÀ ĐỔI TÊN CỘT ====")
df['city-mpg'] = 235 / df['city-mpg']
df.rename(columns={'city-mpg': 'city-L/100km'}, inplace=True)
print(df)
print("\n")

# Ví dụ 2: Kiểm tra và chuyển đổi kiểu dữ liệu cột 'price' bằng astype
print("==== VÍ DỤ 2: ÉP KIỂU 'price' TỪ OBJECT SANG INTEGER ====")
df['price'] = df['price'].astype("int")
print(df)
print("\n--- Kiểu dữ liệu mới sau khi ép kiểu ---")
print(df.dtypes)
```

# **Data Normalization**

## 1. **The Importance of Normalization**
* **Definition:** A data preprocessing technique used to make the range of values consistent across different features.
* **Purpose:** Enables fairer comparison between features, ensuring they have the same impact on statistical models. It is also important for computational reasons.
* **The Problem of Unnormalized Data:** Features with naturally larger values (e.g., income ranging from 20,000 to 500,000) will intrinsically influence results (like linear regression) more heavily than features with smaller values (e.g., age ranging from 0 to 100), incorrectly making them seem like more important predictors.

## 2. **Common Normalization Techniques**
1. **Simple Feature Scaling:** Divides each value by the maximum value for that feature.
   * **Resulting Range:** 0 to 1.
2. **Min-Max:** Takes each value, subtracts the minimum value of the feature, and divides by the range (maximum - minimum).
   * **Resulting Range:** 0 to 1.
3. **Z-score (Standard Score):** Subtracts the mean (mu) from each value and divides by the standard deviation (sigma).
   * **Resulting Range:** Hovers around 0, typically between -3 and +3 (but can extend higher or lower).

## 3. **Applying Normalization in Python (Pandas)**
* **Simple Feature Scaling:** `df['column'] = df['column'] / df['column'].max()`
* **Min-Max:** `df['column'] = (df['column'] - df['column'].min()) / (df['column'].max() - df['column'].min())`
* **Z-score:** `df['column'] = (df['column'] - df['column'].mean()) / df['column'].std()`


```python
import pandas as pd

# Tạo DataFrame với dải dữ liệu length từ 150 đến 250 như ví dụ trong video
df = pd.DataFrame({'length': [150.0, 175.0, 200.0, 225.0, 250.0]})

print("==== DỮ LIỆU BAN ĐẦU ====")
print(df)
print("\n")

# Phương pháp 1: Simple Feature Scaling (Chia cho giá trị lớn nhất)
df_simple = df.copy()
df_simple['length'] = df_simple['length'] / df_simple['length'].max()
print("==== PHƯƠNG PHÁP 1: SIMPLE FEATURE SCALING ====")
print(df_simple)
print("\n")

# Phương pháp 2: Min-Max (Trừ cho min, chia cho range)
df_minmax = df.copy()
df_minmax['length'] = (df_minmax['length'] - df_minmax['length'].min()) / (df_minmax['length'].max() - df_minmax['length'].min())
print("==== PHƯƠNG PHÁP 2: MIN-MAX NORMALIZATION ====")
print(df_minmax)
print("\n")

# Phương pháp 3: Z-score (Trừ cho mean, chia cho standard deviation)
df_zscore = df.copy()
df_zscore['length'] = (df_zscore['length'] - df_zscore['length'].mean()) / df_zscore['length'].std()
print("==== PHƯƠNG PHÁP 3: Z-SCORE (STANDARD SCORE) ====")
print(df_zscore)
```
