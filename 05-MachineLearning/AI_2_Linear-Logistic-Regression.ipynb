# **I .Introduction to Regression**

* **Definition:** A type of supervised learning model used to predict or estimate a **continuous target variable** based on one or more explanatory features (independent variables). 
* **Core Concept:** It models the relationship between variables (e.g., predicting a car's CO2 emissions based on its engine size and number of cylinders).

## **1/ Types of Regression**
* **Simple Regression:** Uses a **single** independent variable to estimate a dependent variable. 
  * *Simple Linear Regression:* Assumes a straight-line (linear) relationship.
  * *Simple Nonlinear Regression:* Assumes a curved (nonlinear) relationship.
* **Multiple Regression:** Uses **more than one** independent variable to predict a dependent variable. This can also be modeled linearly or nonlinearly.




## **2/ Common Algorithms**
* **Classical Statistical Models:** Linear regression, Polynomial regression.
* **Modern Machine Learning Models:** Random Forest, XGBoost, k-Nearest Neighbors (KNN), Support Vector Machines (SVM), and Neural Networks.

## **3/ Real-World Applications**
Regression is used across various domains whenever the goal is to forecast a continuous numerical value:
* **Business & Retail:** Forecasting yearly sales based on leads and order history.
* **Real Estate:** Predicting house prices based on size, location, and number of bedrooms.
* **Predictive Maintenance:** Estimating when an industrial machine or automobile will require maintenance before it fails.
* **Finance:** Predicting employment income based on hours, education, and experience.
* **Environment:** Estimating regional rainfall based on meteorological factors or predicting the probability and severity of wildfires.
* **Healthcare:** Predicting the spread of infectious diseases or estimating the likelihood of a patient developing conditions like diabetes or cancer.

# **II. Simple Linear Regression Overview**
* **Definition:** A model that estimates a linear relationship between a **single** independent (predictor) variable and a continuous dependent (target) variable.
* **Example:** Predicting a car's CO2 emissions (target) based solely on its engine size (predictor).

## **1/ The Mathematical Model**
* **Equation:** $\hat{y} = \theta_0 + \theta_1 x_1$
  * $\hat{y}$: The predicted response (target value).
  * $x_1$: The independent predictor variable.
  * $\theta_0$: The y-intercept (also known as the bias coefficient).
  * $\theta_1$: The slope coefficient for the predictor variable.
* **Calculation:** The coefficients ($\theta_0$ and $\theta_1$) are calculated mathematically using the means ($\bar{x}$, $\bar{y}$) of the dataset. Once solved, making a prediction is as simple as plugging a new input value into the linear equation.

## **2/ Error Measurement & OLS**
* **Residual Error:** The vertical discrepancy between an actual data point and the predicted value on the fitted regression line.
* **Mean Squared Error (MSE):** The average of all residual errors, which measures how poorly the regression line fits the data.
* **Ordinary Least Squares (OLS):** The standard mathematical method used in linear regression. Its primary goal is to find the "best-fit line" by minimizing the mean of all residual errors.

## **3/ Pros and Cons of OLS Regression**
* **Advantages:** It is highly interpretable, easy to understand, requires no tuning, and computes very quickly, especially on small datasets.
* **Disadvantages:** The linear model is too simplistic to capture complex, nonlinear relationships in data. Furthermore, it is highly sensitive to outliers, which can skew the calculations and drastically reduce accuracy.

# **III. Multiple Linear Regression Overview**
* **Definition:** An extension of simple linear regression that uses **two or more** independent variables (features) to estimate a continuous dependent variable (target).
* **Mathematical Model:** Represented as a linear combination of features and weights:

$\hat{y} = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + \dots + \theta_n x_n$

  * y_hat: Predicted response.
  * x_i: Feature vectors (independent variables).
  * theta_0: Bias or intercept term.
  * theta_i: Unknown weights (coefficients) assigned to each feature.
* **Geometric Representation:** While simple linear regression forms a line, multiple linear regression forms a **plane** (for two features) or a **hyperplane** (for more than two features).

## **1/ Handling Data Types**
* **Categorical Variables:** Must be converted into numerical formats to be used in the model. 
  * *Binary variables:* Mapped to 0 and 1 (e.g., manual = 0, automatic = 1).
  * *Multi-class variables:* Transformed into new Boolean features (one-hot encoding) for each class.

## **2/ Error Measurement & Parameter Estimation**
* **Mean Squared Error (MSE):** The model aims to find the best parameters (theta) by minimizing the average of the squared residual errors (the difference between actual and predicted values).
* **Estimation Methods:**
  * **Ordinary Least Squares (OLS):** Uses linear algebra and matrices to calculate optimal weights directly.
  * **Optimization (Gradient Descent):** Iteratively minimizes the model's error starting with random values. This is the preferred approach for very large datasets.

## **3/ Applications & "What-If" Scenarios**
* **Industry Use Cases:** Predicting student exam performance based on revision time and attendance, or forecasting health metrics (like blood pressure) based on changes in BMI.
* **What-If Scenarios:** Involve hypothetically changing one input feature while holding others constant to observe the predicted outcome. 

## **4/ Pitfalls & Limitations**
* **Overfitting:** Adding too many variables can cause the model to memorize the training data, making it a poor predictor for new, unseen observations.
* **Collinearity (Correlated Variables):** If two independent variables are highly correlated (they predict each other), they distort the model and make "what-if" scenarios impossible. **Solution:** Remove redundant, correlated variables and keep only those most controllable and highly correlated with the target.
* **Extrapolation:** Making predictions based on impossible scenarios or data values that fall too far outside the range of the original training data will yield inaccurate results.

# **IV. Polynomial & Nonlinear Regression Overview**
* **Nonlinear Regression:** A statistical method used when the relationship between variables cannot be captured by a straight line. It uses nonlinear equations (polynomial, exponential, logarithmic, or sinusoidal) to model complex patterns.
* **Underfitting vs. Overfitting:** A simple straight line often underfits real-world data. Conversely, a model with an excessively high degree of complexity (passing through every single data point) overfits, memorizing random noise rather than true underlying trends.

## **1/ Polynomial Regression**
* **Definition:** A special form of nonlinear regression that expresses a nonlinear dependence on input features but maintains a linear dependence on regression coefficients. 
* **Mechanism:** It indirectly fits data to polynomial expressions. For example, a cubic (3rd degree) polynomial equation:
  y = theta_0 + theta_1 * x + theta_2 * x^2 + theta_3 * x^3
* **Linearization:** By substituting variables (e.g., x_1 = x, x_2 = x^2, x_3 = x^3), the model transforms into a linear combination:
  y = theta_0 + theta_1 * x_1 + theta_2 * x_2 + theta_3 * x_3
  Because it is linearized, it can be solved using ordinary multiple linear regression.


> Polynomial regression fits a curved line or surface to data when a straight line does not work well.

## **2/ Common Nonlinear Relationships**
* **Exponential:** Used for patterns with increasing growth rates over time, such as compound interest or China's GDP growth. Example equation: y_hat = theta_0 + theta_1 * e^x.
* **Logarithmic:** Used for patterns demonstrating the law of diminishing returns, where incremental gains slow down over time (e.g., human productivity plateauing after consecutive hours of work).
* **Periodicity:** Uses sinusoidal functions to model repeating seasonal variations (e.g., monthly rainfall or temperature).

## **3/ Model Selection & Optimization**
* **Visual Analysis:** Analyze scatter plots of the target variable against input variables to identify if the pattern resembles a specific mathematical function (linear, exponential, etc.).
* **Parameter Optimization:** If a mathematical expression is established, optimization techniques like gradient descent can be used to find the best parameters.
* **Algorithmic Alternatives:** If a specific mathematical equation is not apparent, alternative Machine Learning algorithms can model the nonlinear data, including:
  * Regression Trees & Random Forests
  * Neural Networks
  * Support Vector Machines (SVM)
  * Gradient Boosting Machines
  * k-Nearest Neighbors (KNN)

# **V. Introduction to Logistic Regression**

* **Definition:** A statistical modeling technique used primarily as a **binary classifier**. Despite the word "regression" in its name, it is used for **classification by predicting the probability** that an observation belongs to one of two classes (e.g., True/False, 1/0, Yes/No).
* **Core Concept:** It acts as both a **probability predictor** and a **binary classifier**.



## **1/ When to Use Logistic Regression**
* **Binary Target:** The target variable has exactly two possible outcomes (0 or 1).
* **Probabilistic Need:** You need to know the actual probability of an outcome (e.g., an 80% chance a customer will buy a product), not just a hard "Yes" or "No".
* **Feature Importance:** You want to understand the impact of independent features (variables) by looking at the size of their model coefficients (weights).
* **Linearly Separable Data:** The decision boundary separating the classes can be drawn as a line, plane, or hyperplane.

## **2/ Real-World Applications**
* **Telecommunications:** Predicting customer churn (whether a customer will leave or stay next month).
* **Healthcare:** Estimating the probability of a heart attack or the likelihood of developing diabetes based on characteristics like age, weight, and blood tests.
* **Finance:** Predicting the likelihood of a homeowner defaulting on a mortgage.
* **Engineering:** Estimating the probability of failure for a system or product.

## **3/ The Math: Why not Linear Regression?**
* **The Problem with Linear Regression:** A linear equation ($\hat{y} = \theta_0 + \theta_1 x_1$) generates values that extend infinitely in both positive and negative directions. Probabilities, however, must be strictly contained between **0 and 1**. 
* **The Step Function Flaw:** Simply using a rigid threshold (e.g., if $\hat{y} > 0.5$ then 1, else 0) is too abrupt. It fails to output a smooth probability and treats all values above the threshold identically, regardless of magnitude.

## **4/ The Solution: Sigmoid (Logit) Function**
* **The Sigmoid Function:** Converts any continuous input into a value strictly between 0 and 1, creating a smooth probability curve (S-curve). 
  * Equation: $\sigma(x) = \frac{1}{1 + e^{-x}}$
  * When $x = 0$, $\sigma(x) = 0.5$
  * As $x$ increases, the output approaches 1. As $x$ decreases, the output approaches 0.
* **Predicted Probability ($\hat{p}$):** The model applies the sigmoid function to the linear regression output to predict the probability that the observation belongs to Class 1. 
  * Note: The probability of belonging to Class 0 is simply $1 - \hat{p}$.
* **Decision Boundary:** A threshold (usually 0.5) is applied to the sigmoid output. If the predicted probability is $> 0.5$, the observation is classified as 1; otherwise, it is classified as 0.

# **VI. Training a Logistic Regression Model**
* **Objective:** Find the optimal set of parameters ($\theta$) that map input features to target outcomes with minimal prediction error.
* **The Training Cycle:**
  1. Initialize a starting set of parameters ($\theta$), often randomly.
  2. Predict the probability that the class is 1 for each observation in the data.
  3. Measure the error between predicted probabilities and actual classes using a cost function.
  4. Update the parameters ($\theta$) to reduce this error.
  5. Repeat the process iteratively until the error drops to a satisfactory level or a maximum number of iterations is reached.

## **1/ The Cost Function: Log Loss**
* **Definition:** The specific metric used to evaluate and optimize logistic regression models. It measures how well the predicted probabilities ($\hat{p}_i$) align with the actual binary classes ($y_i$).
* **Mechanism:** 
  * It takes the negative average of logarithmic errors across the dataset.
  * It **rewards** confident and correct classifications by assigning a very small loss (approaching 0).
  * It heavily **penalizes** confident but incorrect predictions, resulting in a massively inflated log loss.

## **2/ Optimization: Gradient Descent**
* **Concept:** An iterative optimization algorithm that adjusts parameter values in the direction of the "steepest descent" (the negative of the gradient) to find the minimum of the cost function.
* **Learning Rate:** A predefined factor that scales the gradient to control the step size. It dictates how far the parameters adjust on each iteration.
* **Limitation:** Standard Gradient Descent calculates the gradient using the **entire dataset** on every single iteration. This makes it computationally heavy and extremely slow for large datasets.

## **3/ Stochastic Gradient Descent (SGD)**
* **Definition:** A variation of standard Gradient Descent that approximates the cost function gradient using a **random subset** (mini-batch) of the training data rather than the entire dataset.
* **Advantages:** It is significantly faster, scales well to massive datasets, and is more capable of bypassing "local minima" to find the "global minimum."
* **Convergence Behavior:** Because it relies on random subsets, SGD tends to wander or bounce around the global minimum rather than settling immediately. 
* **Tuning SGD:** To force the model to settle accurately at the minimum, practitioners can gradually decrease the learning rate or progressively increase the size of the random data subset as the algorithm nears the optimum.
