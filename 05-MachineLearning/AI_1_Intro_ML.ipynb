# **OVERVIEW**
## **AI & Machine Learning Overview**
* **Artificial Intelligence (AI):** A broad field that makes computers appear intelligent by simulating human cognitive abilities. Includes computer vision, NLP, generative AI, machine learning, and deep learning.
* **Machine Learning (ML):** A subset of AI that uses algorithms to learn directly from data, identify patterns, and make decisions without explicit programming. It requires *feature engineering* by practitioners.
* **Deep Learning:** Distinguishes itself from traditional ML by using multi-layered neural networks to automatically extract features from highly complex, unstructured big data.

## **Machine Learning Approaches**
* **Supervised Learning:** Trains on *labeled data* to learn how to make inferences and predict unknown labels for new data.
* **Unsupervised Learning:** Works with *unlabeled data* to discover hidden patterns and groupings.
* **Semi-supervised Learning:** Trains on a small subset of labeled data, then iteratively retrains itself by generating new labels for the rest of the data with high confidence.
* **Reinforcement Learning:** Simulates an AI agent interacting with its environment, learning to make decisions based on feedback (rewards/penalties).

## **Core ML Techniques**
* **Classification:** Categorizes input data into predefined classes (e.g., predicting if a cancer cell is benign or malignant, or if a customer will churn).
* **Regression / Estimation:** Predicts continuous numerical values (e.g., house prices, engine CO2 emissions).
* **Clustering:** An unsupervised technique that groups similar data points together (e.g., customer segmentation). Uncategorized data points are considered noise.
* **Association:** Finds items or events that frequently co-occur (e.g., grocery items bought together).
* **Anomaly Detection:** Discovers abnormal and unusual cases (e.g., credit card fraud detection).
* **Sequence Mining:** Predicts the next event in a sequence (e.g., website clickstream analytics).
* **Dimension Reduction:** Reduces data size by decreasing the number of required features.
* **Recommendation Systems:** Associates people's preferences with others who have similar tastes to recommend new items (e.g., Netflix movies, Amazon products).

## **Real-World Applications**
* **Healthcare:** Predicting whether a cell is benign or malignant based on characteristics like clump thickness and uniformity.
* **Banking & Finance:** Predicting applicants' default probabilities to approve/deny loans.
* **Telecommunications:** Segmenting demographic data to predict customer churn.
* **Computer Vision:** Building models that automatically learn distinguishing features to identify objects (e.g., cats vs. dogs) instead of painstakingly coding fixed rules.

# **LIFECYCLE**
## **The Machine Learning Model Lifecycle in Practice**

* **1/ Problem Definition:** Start by understanding the client's needs and defining the end-user's pain points. This ensures the ML solution aligns with business goals (e.g., building a recommendation system to increase revenue and address customers' specific skincare needs).
* **2/ Data Collection:** Identify and gather data from multiple sources. This includes user data (demographics, purchase history, saved items) and product data (ingredients, ratings, popularity). The data is then wrangled, merged, and mapped into a single central source.
* **3/ Data Preparation (often overlaps with Collection):** Focus on cleaning and finalizing the dataset for the model. 
  * Filter irrelevant data, remove or handle extreme outliers, and address missing values.
  * Ensure proper column formatting (e.g., correctly parsing dates and strings).
  * **Feature Engineering:** Create new variables, such as calculating the average time between transactions or mapping specific skin issues to users.
  * **Exploratory Data Analysis (EDA):** Use plots and correlation analysis to identify patterns and validate data with subject matter experts.
  * **Data Splitting:** Determine the strategy for splitting the data into training and testing sets (e.g., reserving the most recent transaction for the test set).
* **4/ Model Development:** Leverage existing frameworks rather than building from scratch. For a recommendation system, techniques often combined include:
  * **Content-Based Filtering:** Recommends items based on product similarity (e.g., matching a cleanser's ingredients to a highly moisturizing lotion).
  * **Collaborative Filtering:** Recommends items based on user similarity (grouping users by characteristics and suggesting what similar users rated highly).
* **5/ Model Evaluation:** Test and tune the model using the pre-defined test dataset. Conduct real-world experiments by deploying to a small user group to collect feedback and track business metrics like click-through rates and purchases.
* **6/ Model Deployment:** Integrate the finalized model into the production environment (e.g., a web or mobile app). Continuous monitoring is essential to track its real-world performance, and the model must be iteratively retrained with new data to expand capabilities and maintain quality.

## **Key Concepts**
* **Iterative Nature:** The ML lifecycle is not strictly linear. Practitioners frequently move back and forth between stages. For example, if issues arise with a deployed model in production, it may require revisiting the problem definition or data collection stages and repeating the process.
* **ETL (Extract, Transform, and Load):** Refers directly to the Data Collection and Data Preparation stages. It involves extracting data from multiple sources, transforming and cleaning it, and loading it into a centralized, accessible repository so ML engineers can build models.

# **TOOLS**
## **1/ Data Processing & Analytics Tools**
* **PostgreSQL:** An open-source object-relational database system based on SQL.
* **Hadoop:** A highly scalable, disk-based solution for storing and batch-processing massive data.
* **Spark:** A distributed, in-memory framework for real-time big data processing (faster than Hadoop, supports DataFrames/SQL).
* **Apache Kafka:** A distributed streaming platform for big data pipelines and real-time analytics.
* **Pandas:** A Python library for exploring, cleaning, and wrangling structured data using DataFrames.
* **NumPy:** A Python library providing mathematical functions, random number generators, and linear algebra routines for distributed and GPU-based computing.

## **2/ Data Visualization Tools**
* **Matplotlib:** A foundational Python library for generating customizable plots and interactive visualizations.
* **Seaborn:** Built on Matplotlib, providing a high-level interface for attractive statistical graphics.
* **ggplot2:** An open-source R package that allows building and adding elements to graphics in layers.
* **Tableau:** A business intelligence tool for interactive data visualization dashboards.

## **3/ Classical Machine Learning Ecosystem**
* **NumPy:** Provides foundational support with efficient numerical computations on large, multidimensional arrays.
* **SciPy:** Built on NumPy for scientific computing, offering modules for optimization, integration, and linear regression.
* **Scikit-learn:** Built on NumPy, SciPy, and Matplotlib. Offers a full suite of classical ML algorithms for classification, regression, clustering, and dimensionality reduction.

## **4/ Deep Learning Frameworks**
* **TensorFlow:** An open-source library for numerical computing and large-scale ML.
* **Keras:** An easy-to-use library for implementing neural networks.
* **Theano:** Efficiently defines, optimizes, and evaluates mathematical expressions involving arrays.
* **PyTorch:** An open-source library for deep learning, computer vision, NLP, and rapid experimentation.

## **5/ Computer Vision Tools**
* **OpenCV:** Real-time CV applications like object detection, image classification, and augmented reality.
* **Scikit-Image:** Built on SciPy and compatible with Pandas; offers image processing algorithms like filters, segmentation, and feature extraction.
* **TorchVision:** Part of PyTorch; provides datasets, image loading, pre-trained architectures, and common image transformations.

## **6/ Natural Language Processing (NLP) Tools**
* **NLTK (Natural Language Toolkit):** Offers text processing, tokenization, and stemming.
* **TextBlob:** Used for part-of-speech tagging, noun-phrase extraction, sentiment analysis, and translation.
* **Stanza:** Stanford's library with accurate pre-trained models for tasks like named entity recognition and dependency parsing.

## **7/ Generative AI Tools**
* **Hugging Face Transformers:** A library of transformer models for text generation, translation, and sentiment analysis.
* **ChatGPT:** A language model for text generation and chatbots.
* **DALL-E:** An OpenAI tool for generating images from textual descriptions.
* **PyTorch:** Can be used to create generative models like GANs and Transformers for text/image generation.

## **8/ Common ML Programming Languages**
* Python, R, Julia, Scala, Java, and JavaScript.
