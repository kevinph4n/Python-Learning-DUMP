## **METHOD**
### 1/ .upper()

```python
A = "i can't help it but fall in love with you"
B = A.upper()
print(B)
```

### 2/ .replace()

```python
A = "kevinph4n is handsome-ish"
B = A.replace('handsome-ish', 'handsome')
print(B)
```

### 3/ .find()    |   The output would be the first index of the string

```python
name = "Kevin Phan"
A = name.find('Phan')
B = name.find('Kevin')
print(A)
print(B)
```

### 4/ .title()

```python
name = "im a beast iykyk"
A = name.title()
print(A)
```
