### **1/ `def` to define functions**

```python
# define the function

print("in ra hello world o tren, python o duoi:\n")

def hello():
    print("Hello World")
    print("Python")

# function calling
hello()
```

```python
# with parameters

print("1/ in ra name:\n")

def hi(name):
    print(f"hi {name}")

hi("kevin")   
hi("hdan")   

print("\n2/ lam phep cong:\n")

def sum(a, b):
    print(f"{a} + {b} = {a + b}")

sum(67, 41)    
sum(69, 21)    
```
