### Pandas in Python (Data Science)
- pandas is a software library built for the Python programming language used to handle, clean, and analyze data.
- What it does: It organizes information into rows and columns, similar to a spreadsheet or an SQL table.
- Main uses: Developers and data scientists use it to filter data, fill in missing values, group numbers together, and prepare data for machine learning.
- Core features: It relies on fast data structures called Series (one-dimensional) and DataFrames (two-dimensional).

```python
# u can call it whatever u want
# you can now have access to pre-built classes and functions such as:
# read_csv()
# Series()
# DataFrame
# Values
import pandas as pd
# Create a Series from a list
data = [10, 20, 30, 40, 50]
s = pd.Series(data)
print(s)
```

### DataFrames?
- a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure. Think of it exactly like a spreadsheet, SQL table, or an Excel sheet where data is organized into labeled rows and columns.
- can be created from dictionaries, with keys as column labels and values as lists representing rows.

```python
import pandas as pd

# Creating a DataFrame from a dictionary
dataFrame = {'Name': ['Kevin', 'Nvi', 'kate', 'David'],
        'Age': [18, 20, 19, 28],
        'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']}

df = pd.DataFrame(dataFrame)
# df means everything
print(df)
```

```python
import pandas as pd

dataFrame = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Score": [8.5, 9.0, 7.5]
})
print(dataFrame)
```
